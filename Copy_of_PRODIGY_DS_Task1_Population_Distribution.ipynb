```python
import pandas as pd

df = pd.read_csv("API_SP.POP.TOTL_DS2_en_csv_v2_33112.csv", skiprows=4)

df.head()
```

```python
df.columns
```

```python
df[["Country Name", "2020"]].head(10)
```

```python
import matplotlib.pyplot as plt

plt.hist(df["2020"].dropna(), bins=10)

plt.title("Distribution of Population in 2020")
plt.xlabel("Population")
plt.ylabel("Number of Countries")

plt.show()
```

```python
import matplotlib.pyplot as plt

population_billions = df["2020"].dropna() / 1_000_000_000

plt.figure(figsize=(10, 6))

plt.hist(population_billions, bins=10)

plt.title("Distribution of Population Across Countries in 2020")
plt.xlabel("Population (Billions)")
plt.ylabel("Number of Countries")

plt.show()
```

Observation: The histogram shows the distribution of population across countries in 2020. The distribution is highly right-skewed, indicating that most countries have relatively smaller populations, while only a few countries have very large populations.
